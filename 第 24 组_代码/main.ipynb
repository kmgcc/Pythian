## 3. 特征工程与光谱 PCA

模型输入共 12 个原始字段：时间特征（`hour`、`month`）、太阳位置（`solar_altitude`、`solar_azimuth`）、天气特征（`cloud_cover`、`humidity`、`temperature`、`precipitation`）、观测辅助（`outdoor_lux`，即水平照度）、类别特征（`weather`、`sky_condition`、`location_code`，独热编码）。

模型输出是 41 维相对光谱。每条实测光谱除以自身最大值归一化到 0-1，模型只学习光谱形状，照度高低由 `outdoor_lux` 等特征单独表达。

预处理统一放进 sklearn 的 `ColumnTransformer`：数值特征标准化（KNN、MLP、线性模型都依赖），类别特征独热编码。

```python
from src.preprocessing import (
    NUMERIC_FEATURES, CATEGORICAL_FEATURES, FEATURE_COLUMNS,
    extract_features_targets, split_dataset, build_preprocessor,
)

print("数值特征:", NUMERIC_FEATURES)
print("类别特征:", CATEGORICAL_FEATURES)

x, y = extract_features_targets(dataset)
print(f"\n特征表 X: {x.shape}，相对光谱目标 Y: {y.shape}")
display(x.head())
print("独热编码后的特征维度:", build_preprocessor().fit_transform(x).shape[1])
```

光谱相邻波长高度相关，直接预测 41 维既慢又难解释。先用 10 个主成分观察解释方差衰减情况，再决定保留多少个。

```python
from src.spectrum_pca import SpectrumPCA
from src.visualization import plot_pca_variance

FIG_DIR = "outputs/figures"

# 先用 10 个主成分观察解释方差衰减情况
pca_probe = SpectrumPCA(n_components=10)
pca_probe.fit_transform(y)
plot_pca_variance(pca_probe.pca, save_path=f"{FIG_DIR}/pca_variance.png")
plt.show()
display(pca_probe.variance_frame())
```

前 5 个主成分累计解释方差约 98%，之后每个主成分的贡献都低于 1%，继续增加对还原精度帮助很小，反而增加模型输出维度。保留 **5 个主成分**。

```python
N_COMPONENTS = 5
spectrum_pca = SpectrumPCA(n_components=N_COMPONENTS)
y_pca = spectrum_pca.fit_transform(y)
print(f"PCA 降维: {y.shape[1]} 维光谱 → {y_pca.shape[1]} 个主成分系数")
print(f"累计解释方差: {spectrum_pca.explained_variance_ratio_.sum():.4f}")

# 划分训练集/测试集（按天气类别分层）
x_train, x_test, y_train_pca, y_test_pca, idx_train, idx_test = split_dataset(x, y_pca)
y_test_spectrum = y[idx_test]   # 测试集真实相对光谱（用于在光谱空间统一评价）
print(f"训练集 {len(x_train)} 条，测试集 {len(x_test)} 条")
```

## 6. 模型参数调优与结果评估

`n_estimators=180, max_depth=18, min_samples_leaf=2` 只是一组经验参数，不一定是最优组合。用 `GridSearchCV` 做系统化调参：

**调参策略：**
1. 在 PCA 空间做 5 折交叉验证（效率高、避免过拟合评估）
2. 对每组参数，在全量训练集上拟合、测试集上预测，逆变换回光谱空间计算 MAE / RMSE / R²
3. 综合 RMSE（主指标）和训练时间（效率指标）选出最优参数
4. 用最优参数训练最终模型

**参数网格：**
- `n_estimators`: [100, 200, 300] —— 树的数量
- `max_depth`: [10, 20, None] —— 树的最大深度
- `min_samples_split`: [2, 5, 10] —— 内部节点再分裂所需最小样本数

共 3 × 3 × 3 = 27 组参数组合。

```python
from src.hyperparameter_tuning import tune_random_forest, tuning_summary_text

# 随机森林超参数调优：27 组参数 × 5 折交叉验证
tuning_result = tune_random_forest(
    x_train=x_train,
    y_train_pca=y_train_pca,
    x_test=x_test,
    y_test_pca=y_test_pca,
    y_test_spectrum=y_test_spectrum,
    spectrum_pca=spectrum_pca,
    cv=5,
    verbose=True,
)

print("\n" + tuning_summary_text(tuning_result))
```

```python
# 展示全部参数组合的评估结果（按 RMSE 升序）
tuning_df = tuning_result.summary_frame()
display(tuning_df)
```

**调优结论：** `n_estimators` 在 100 棵时已收敛、`max_depth=10` 明显欠拟合而 20 与不限深度无差别、`min_samples_split=5` 略优于默认值 2。综合 RMSE 最小和训练效率，最终选出 `n_estimators=100, max_depth=20, min_samples_split=5`，训练 0.17s、预测 0.016s，测试集 RMSE=0.0090、MAE=0.0051、R²=0.9513，比原始经验值（180, 18, 2）更精简高效。下面用该参数训练最终模型并保存，后续章节使用调优后的模型。

```python
# 用调优后的最优参数训练最终随机森林模型
best_rf_pipeline = tuning_result.best_estimator
best_rf_params = tuning_result.best_params

print("调优后最优参数:")
for k, v in best_rf_params.items():
    print(f"  {k}: {v}")

# 重新预测并更新后续章节使用的变量
y_pred_pca_best = best_rf_pipeline.predict(x_test)
y_pred_spectrum_best = spectrum_pca.inverse_transform(y_pred_pca_best)

# 更新 trained 字典中的 Random Forest 条目
from src.model_training import TrainedModel
import time as _time

t0 = _time.perf_counter()
y_pred_pca_recheck = best_rf_pipeline.predict(x_test)
predict_time_best = _time.perf_counter() - t0

trained["Random Forest (调优)"] = TrainedModel(
    name="Random Forest (调优)",
    pipeline=best_rf_pipeline,
    y_pred_spectrum=y_pred_spectrum_best,
    train_seconds=tuning_result.best_candidate.train_time_s,
    predict_seconds=predict_time_best,
)

# 更新最佳模型引用
best_name = "Random Forest (调优)"
pred_spectrum = y_pred_spectrum_best[0]

print(f"\n调优后模型已保存到 trained['Random Forest (调优)']")
print(f"后续章节将使用调优后的模型进行预测和补偿。")
```

## 8. 照明补偿应用展示

本节是应用层演示：用户设定一组天气/环境参数，系统走完整条应用链路——模型预测 PCA 主成分系数，逆变换还原自然光相对光谱，再做多通道 LED 补偿并与传统双色温 LED 对照，最后对比光谱与色度。

涉及的应用层算法：

目标光谱取实测晴天高太阳高度角日光的平均相对光谱（由真实数据计算，不是人工设定曲线）；当前光谱是模型预测的自然光相对光谱乘以室内自然光占比（由室外照度和目标照度估算）。多通道补偿用非负最小二乘求解七个 LED 通道比例，使自然光加 LED 混光尽量接近目标光谱。传统双色温对照则是在冷暖配比和亮度两个旋钮下搜索与目标误差最小的组合，说明"能调冷暖，但光谱形状受限"。近似色度预览用 CIE 2015（CIE 2006 生理学数据推导）10° 观察者色匹配函数积分得 XYZ，亮度归一化后转 sRGB，刻意不使用 1931 年的旧标准观察者。

四个天气预设不是手写参数：每个预设先在真实数据集中筛出对应场景子集，数值特征取中位数、类别特征取众数，保证输入和训练数据同分布。

```python
from src.led_spectrum_data import fetch_and_build_dual_white_spectrum
from src.model_training import SpectrumPredictor
from src.preprocessing import FEATURE_COLUMNS
from src.application_demo import (
    PRESET_DEFINITIONS, PRESET_NAMES, run_application_demo,
    pca_coefficient_frame, channel_percent_frame,
    error_comparison_frame, color_preview_frame, preset_summary_frame,
)

# 确保实测 LED 光谱就绪（已缓存则直接复用）
fetch_and_build_dual_white_spectrum(force=False)

# 把第 4-5 节训练好的最佳模型包装成预测器，供应用层调用
predictor = SpectrumPredictor(
    best_model_name=best_name,
    pipeline=trained[best_name].pipeline,
    spectrum_pca=spectrum_pca,
    feature_columns=FEATURE_COLUMNS,
    metrics=metrics,
)

print("可选天气预设：")
for name in PRESET_NAMES:
    print(f"  - {name}：{PRESET_DEFINITIONS[name]['说明']}")

PRESET = "晴天中午"   # 课堂演示时改这里即可切换场景
demo = run_application_demo(predictor, dataset, preset=PRESET, scene="学习", target_lux=500.0)

print(f"\n当前预设「{PRESET}」生成的模型输入：")
display(demo.features)
```
